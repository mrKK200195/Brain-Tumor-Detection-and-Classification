```python
#CNN Baseline


from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf

from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

# SETTINGS

IMG_SIZE = (224, 224)
BATCH_SIZE = 32
SEED = 42
EPOCHS = 20
LEARNING_RATE = 0.001


print("TensorFlow version:")
print(tf.__version__)

# FIND PROJECT ROOT

CURRENT_DIR = Path.cwd()

PROJECT_ROOT = None

for path in [CURRENT_DIR] + list(CURRENT_DIR.parents):

    if (path / "data" / "raw").exists():

        PROJECT_ROOT = path
        break


if PROJECT_ROOT is None:

    raise FileNotFoundError(
        "Could not find project root containing data/raw."
    )


DATA_ROOT = PROJECT_ROOT / "data" / "raw"

MODELS_DIR = PROJECT_ROOT / "models" / "cnn"
FIGURES_DIR = PROJECT_ROOT / "results" / "figures"
RESULTS_DIR = PROJECT_ROOT / "results"


MODELS_DIR.mkdir(
    parents=True,
    exist_ok=True
)

FIGURES_DIR.mkdir(
    parents=True,
    exist_ok=True
)

RESULTS_DIR.mkdir(
    parents=True,
    exist_ok=True
)


print("\nProject root:")
print(PROJECT_ROOT)

# FIND DATASET FOLDERS


def find_folder(root, folder_name):

    matches = [
        path
        for path in root.rglob(folder_name)
        if path.is_dir()
    ]

    if not matches:

        raise FileNotFoundError(
            f"Could not find '{folder_name}' inside:\n{root}"
        )

    return matches[0]


TRAIN_DIR = find_folder(
    DATA_ROOT,
    "Training"
)

TEST_DIR = find_folder(
    DATA_ROOT,
    "Testing"
)


print("\nTraining directory:")
print(TRAIN_DIR)

print("\nTesting directory:")
print(TEST_DIR)

# CLASS NAMES

CLASS_NAMES = sorted([

    folder.name

    for folder in TRAIN_DIR.iterdir()

    if folder.is_dir()

])


NUM_CLASSES = len(CLASS_NAMES)


print("\nClasses:")
print(CLASS_NAMES)

print("\nNumber of classes:")
print(NUM_CLASSES)


# LOAD DATASETS

train_dataset = tf.keras.utils.image_dataset_from_directory(

    TRAIN_DIR,

    labels="inferred",

    label_mode="categorical",

    class_names=CLASS_NAMES,

    image_size=IMG_SIZE,

    batch_size=BATCH_SIZE,

    validation_split=0.20,

    subset="training",

    seed=SEED,

    shuffle=True

)


validation_dataset = tf.keras.utils.image_dataset_from_directory(

    TRAIN_DIR,

    labels="inferred",

    label_mode="categorical",

    class_names=CLASS_NAMES,

    image_size=IMG_SIZE,

    batch_size=BATCH_SIZE,

    validation_split=0.20,

    subset="validation",

    seed=SEED,

    shuffle=True

)


test_dataset = tf.keras.utils.image_dataset_from_directory(

    TEST_DIR,

    labels="inferred",

    label_mode="categorical",

    class_names=CLASS_NAMES,

    image_size=IMG_SIZE,

    batch_size=BATCH_SIZE,

    shuffle=False

)


# PREFETCH DATA

AUTOTUNE = tf.data.AUTOTUNE


train_dataset = train_dataset.prefetch(
    AUTOTUNE
)

validation_dataset = validation_dataset.prefetch(
    AUTOTUNE
)

test_dataset = test_dataset.prefetch(
    AUTOTUNE
)


# DATA AUGMENTATION

data_augmentation = tf.keras.Sequential([

    tf.keras.layers.RandomRotation(
        0.08
    ),

    tf.keras.layers.RandomZoom(
        0.10
    ),

    tf.keras.layers.RandomTranslation(
        0.05,
        0.05
    )

], name="data_augmentation")

#BUILD CNN MODEL

model = tf.keras.Sequential([

    tf.keras.layers.Input(
        shape=(224, 224, 3)
    ),

    # Data augmentation

    data_augmentation,

    # -------------------------
    # Normalize pixels
    # -------------------------

    tf.keras.layers.Rescaling(
        1.0 / 255.0
    ),

    # CONVOLUTIONAL BLOCK 1
    

    tf.keras.layers.Conv2D(
        32,
        (3, 3),
        padding="same"
    ),

    tf.keras.layers.BatchNormalization(),

    tf.keras.layers.Activation(
        "relu"
    ),

    tf.keras.layers.MaxPooling2D(
        (2, 2)
    ),

    # CONVOLUTIONAL BLOCK 2

    tf.keras.layers.Conv2D(
        64,
        (3, 3),
        padding="same"
    ),

    tf.keras.layers.BatchNormalization(),

    tf.keras.layers.Activation(
        "relu"
    ),

    tf.keras.layers.MaxPooling2D(
        (2, 2)
    ),

    tf.keras.layers.Dropout(
        0.20
    ),


    # CONVOLUTIONAL BLOCK 3

    tf.keras.layers.Conv2D(
        128,
        (3, 3),
        padding="same"
    ),

    tf.keras.layers.BatchNormalization(),

    tf.keras.layers.Activation(
        "relu"
    ),

    tf.keras.layers.MaxPooling2D(
        (2, 2)
    ),

    tf.keras.layers.Dropout(
        0.25
    ),

    # CONVOLUTIONAL BLOCK 4

    tf.keras.layers.Conv2D(
        256,
        (3, 3),
        padding="same"
    ),

    tf.keras.layers.BatchNormalization(),

    tf.keras.layers.Activation(
        "relu"
    ),

    tf.keras.layers.MaxPooling2D(
        (2, 2)
    ),

    tf.keras.layers.Dropout(
        0.30
    ),

    # CLASSIFICATION HEAD

    tf.keras.layers.GlobalAveragePooling2D(),

    tf.keras.layers.Dense(
        256,
        activation="relu"
    ),

    tf.keras.layers.Dropout(
        0.50
    ),

    tf.keras.layers.Dense(
        NUM_CLASSES,
        activation="softmax"
    )

], name="CNN_Baseline")

# 9. DISPLAY MODEL

print("\nCNN MODEL SUMMARY")
print("=" * 60)

model.summary()

# 10. COMPILE MODEL

model.compile(

    optimizer=tf.keras.optimizers.Adam(
        learning_rate=LEARNING_RATE
    ),

    loss="categorical_crossentropy",

    metrics=[
        "accuracy"
    ]

)

# 11. CALLBACKS

best_model_path = (
    MODELS_DIR /
    "best_cnn.keras"
)


early_stopping = tf.keras.callbacks.EarlyStopping(

    monitor="val_loss",

    patience=5,

    restore_best_weights=True

)


reduce_lr = tf.keras.callbacks.ReduceLROnPlateau(

    monitor="val_loss",

    factor=0.5,

    patience=2,

    min_lr=1e-7,

    verbose=1

)


model_checkpoint = tf.keras.callbacks.ModelCheckpoint(

    filepath=str(best_model_path),

    monitor="val_accuracy",

    save_best_only=True,

    mode="max",

    verbose=1

)

# 12. TRAIN MODEL

print("\n")
print("=" * 60)
print("STARTING CNN TRAINING")
print("=" * 60)


history = model.fit(

    train_dataset,

    validation_data=validation_dataset,

    epochs=EPOCHS,

    callbacks=[
        early_stopping,
        reduce_lr,
        model_checkpoint
    ]

)

# TRAINING HISTORY

history_df = pd.DataFrame(
    history.history
)


print("\nTraining history:")
display(history_df)


# PLOT ACCURACY


plt.figure(figsize=(10, 6))


plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)


plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)


plt.xlabel("Epoch")
plt.ylabel("Accuracy")

plt.title(
    "CNN Training and Validation Accuracy"
)

plt.legend()

plt.grid(True)

plt.tight_layout()


accuracy_path = (
    FIGURES_DIR /
    "cnn_accuracy.png"
)


plt.savefig(
    accuracy_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# PLOT LOSS

plt.figure(figsize=(10, 6))


plt.plot(
    history.history["loss"],
    label="Training Loss"
)


plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)


plt.xlabel("Epoch")
plt.ylabel("Loss")

plt.title(
    "CNN Training and Validation Loss"
)

plt.legend()

plt.grid(True)

plt.tight_layout()


loss_path = (
    FIGURES_DIR /
    "cnn_loss.png"
)


plt.savefig(
    loss_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()


#  LOAD BEST MODEL

print("\nLoading best CNN model...")

best_model = tf.keras.models.load_model(
    best_model_path
)


print("Best model loaded successfully.")

#EVALUATE ON TEST DATA

print("\n")
print("=" * 60)
print("EVALUATING CNN ON TEST DATA")
print("=" * 60)


test_loss, test_accuracy = best_model.evaluate(
    test_dataset,
    verbose=1
)


print("\nTest Loss:")
print(test_loss)

print("\nTest Accuracy:")
print(test_accuracy)


# GET PREDICTIONS

y_true = []
y_pred = []


for images, labels in test_dataset:

    predictions = best_model.predict(
        images,
        verbose=0
    )

    true_labels = np.argmax(
        labels.numpy(),
        axis=1
    )

    predicted_labels = np.argmax(
        predictions,
        axis=1
    )

    y_true.extend(
        true_labels
    )

    y_pred.extend(
        predicted_labels
    )


y_true = np.array(y_true)
y_pred = np.array(y_pred)

# CLASSIFICATION METRICS

accuracy = accuracy_score(
    y_true,
    y_pred
)

precision = precision_score(
    y_true,
    y_pred,
    average="weighted",
    zero_division=0
)

recall = recall_score(
    y_true,
    y_pred,
    average="weighted",
    zero_division=0
)

f1 = f1_score(
    y_true,
    y_pred,
    average="weighted",
    zero_division=0
)


print("\n")
print("=" * 60)
print("CNN TEST METRICS")
print("=" * 60)

print(
    f"Accuracy : {accuracy:.4f}"
)

print(
    f"Precision: {precision:.4f}"
)

print(
    f"Recall   : {recall:.4f}"
)

print(
    f"F1 Score : {f1:.4f}"
)

# CLASSIFICATION REPORT

print("\nClassification Report")
print("=" * 60)


report = classification_report(

    y_true,

    y_pred,

    target_names=CLASS_NAMES,

    digits=4,

    zero_division=0

)


print(report)

# CONFUSION MATRIX


cm = confusion_matrix(
    y_true,
    y_pred
)


plt.figure(
    figsize=(8, 7)
)


disp = ConfusionMatrixDisplay(

    confusion_matrix=cm,

    display_labels=CLASS_NAMES

)


disp.plot(
    cmap="Blues",
    values_format="d",
    ax=plt.gca(),
    colorbar=False
)


plt.title(
    "CNN Confusion Matrix"
)

plt.tight_layout()


cm_path = (
    FIGURES_DIR /
    "cnn_confusion_matrix.png"
)


plt.savefig(
    cm_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# SAVE METRICS

metrics_df = pd.DataFrame({

    "Model": [
        "CNN Baseline"
    ],

    "Accuracy": [
        accuracy
    ],

    "Precision": [
        precision
    ],

    "Recall": [
        recall
    ],

    "F1 Score": [
        f1
    ]

})


metrics_path = (
    RESULTS_DIR /
    "cnn_metrics.csv"
)


metrics_df.to_csv(
    metrics_path,
    index=False
)


# SAVE TRAINING HISTORY


history_path = (
    RESULTS_DIR /
    "cnn_training_history.csv"
)


history_df.to_csv(
    history_path,
    index=False
)

# FINAL SUMMARY

print("\n")
print("=" * 60)
print("CNN BASELINE COMPLETED")
print("=" * 60)

print(
    f"Best model saved to:\n{best_model_path}"
)

print(
    f"\nAccuracy: {accuracy:.4f}"
)

print(
    f"Precision: {precision:.4f}"
)

print(
    f"Recall: {recall:.4f}"
)

print(
    f"F1 Score: {f1:.4f}"
)

print(
    f"\nMetrics saved to:\n{metrics_path}"
)

print(
    f"Training history saved to:\n{history_path}"
)

print(
    f"Confusion matrix saved to:\n{cm_path}"
)

print(
    "\nNext notebook:"
)

print(
    "04_transfer_learning.ipynb"
)
```
