```python
# HYPERPARAMETER TUNING + FINAL TUNED CNN

import warnings
warnings.filterwarnings("ignore")

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf

from pathlib import Path

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

from sklearn.utils.class_weight import compute_class_weight

# CONFIGURATION

IMG_SIZE = (224, 224)
BATCH_SIZE = 32
SEED = 42

# Short tuning stage for CPU
EPOCHS = 10
PATIENCE = 3

# Final training
FINAL_EPOCHS = 30
FINAL_PATIENCE = 7

# Previous CNN baseline results
BASELINE_ACCURACY = 0.7450
BASELINE_PRECISION = 0.7372
BASELINE_RECALL = 0.7450
BASELINE_F1 = 0.7337

tf.keras.utils.set_random_seed(SEED)
np.random.seed(SEED)

print("=" * 70)
print("NOTEBOOK 06 - HYPERPARAMETER TUNING")
print("=" * 70)

print()
print("TensorFlow version:", tf.__version__)
print("NumPy version:", np.__version__)
print("Pandas version:", pd.__version__)

print()
print("Configuration loaded.")

# FIND PROJECT ROOT

current_path = Path.cwd().resolve()

PROJECT_ROOT = None

for path in [current_path] + list(current_path.parents):

    if (path / "data" / "raw").exists():

        PROJECT_ROOT = path
        break

if PROJECT_ROOT is None:

    raise FileNotFoundError(
        "Could not find project root containing data/raw"
    )

print()
print("Project root:")
print(PROJECT_ROOT)

# CREATE PROJECT DIRECTORIES

DATA_DIR = PROJECT_ROOT / "data"
RAW_DIR = DATA_DIR / "raw"

RESULTS_DIR = PROJECT_ROOT / "results"
FIGURES_DIR = RESULTS_DIR / "figures"

MODELS_DIR = PROJECT_ROOT / "models"
CNN_MODELS_DIR = MODELS_DIR / "cnn"

RESULTS_DIR.mkdir(
    parents=True,
    exist_ok=True
)

FIGURES_DIR.mkdir(
    parents=True,
    exist_ok=True
)

CNN_MODELS_DIR.mkdir(
    parents=True,
    exist_ok=True
)

print()
print("Output folders ready.")

# FIND TRAINING AND TESTING DIRECTORIES

training_dirs = list(
    RAW_DIR.rglob("Training")
)

testing_dirs = list(
    RAW_DIR.rglob("Testing")
)

if len(training_dirs) == 0:

    raise FileNotFoundError(
        "Training folder was not found."
    )

if len(testing_dirs) == 0:

    raise FileNotFoundError(
        "Testing folder was not found."
    )

TRAIN_DIR = training_dirs[0]
TEST_DIR = testing_dirs[0]

print()
print("Training directory:")
print(TRAIN_DIR)

print()
print("Testing directory:")
print(TEST_DIR)

#  DETECT CLASS NAMES


CLASS_NAMES = sorted(
    [
        folder.name
        for folder in TRAIN_DIR.iterdir()
        if folder.is_dir()
    ]
)

print()
print("Detected classes:")

for i, class_name in enumerate(CLASS_NAMES):

    print(
        f"{i}: {class_name}"
    )

if len(CLASS_NAMES) != 4:

    raise ValueError(
        f"Expected 4 classes, "
        f"but found {len(CLASS_NAMES)}"
    )

# COUNT IMAGES


IMAGE_EXTENSIONS = {
    ".jpg",
    ".jpeg",
    ".png",
    ".bmp",
    ".webp"
}

train_counts = {}
test_counts = {}

for class_name in CLASS_NAMES:

    train_class_dir = (
        TRAIN_DIR / class_name
    )

    test_class_dir = (
        TEST_DIR / class_name
    )

    train_counts[class_name] = sum(
        1
        for file in train_class_dir.rglob("*")
        if file.suffix.lower()
        in IMAGE_EXTENSIONS
    )

    test_counts[class_name] = sum(
        1
        for file in test_class_dir.rglob("*")
        if file.suffix.lower()
        in IMAGE_EXTENSIONS
    )

count_df = pd.DataFrame({

    "Class": CLASS_NAMES,

    "Training": [
        train_counts[c]
        for c in CLASS_NAMES
    ],

    "Testing": [
        test_counts[c]
        for c in CLASS_NAMES
    ]
})

print()
print("Dataset counts:")

display(count_df)

print()
print(
    "Total training images:",
    sum(train_counts.values())
)

print(
    "Total testing images:",
    sum(test_counts.values())
)


# LOAD TRAINING DATASET

print()
print("Loading training dataset...")

train_ds = (
    tf.keras.utils
    .image_dataset_from_directory(

        TRAIN_DIR,

        labels="inferred",

        label_mode="categorical",

        class_names=CLASS_NAMES,

        image_size=IMG_SIZE,

        batch_size=BATCH_SIZE,

        validation_split=0.20,

        subset="training",

        seed=SEED,

        shuffle=True
    )
)

# LOAD VALIDATION DATASET

print()
print("Loading validation dataset...")

val_ds = (
    tf.keras.utils
    .image_dataset_from_directory(

        TRAIN_DIR,

        labels="inferred",

        label_mode="categorical",

        class_names=CLASS_NAMES,

        image_size=IMG_SIZE,

        batch_size=BATCH_SIZE,

        validation_split=0.20,

        subset="validation",

        seed=SEED,

        shuffle=False
    )
)

#LOAD TEST DATASET

print()
print("Loading testing dataset...")

test_ds = (
    tf.keras.utils
    .image_dataset_from_directory(

        TEST_DIR,

        labels="inferred",

        label_mode="categorical",

        class_names=CLASS_NAMES,

        image_size=IMG_SIZE,

        batch_size=BATCH_SIZE,

        shuffle=False
    )
)

# 10. PREFETCH

AUTOTUNE = tf.data.AUTOTUNE

train_ds = train_ds.prefetch(
    AUTOTUNE
)

val_ds = val_ds.prefetch(
    AUTOTUNE
)

test_ds = test_ds.prefetch(
    AUTOTUNE
)

print()
print("Dataset prefetching enabled.")

# DATA AUGMENTATION

augmentation = tf.keras.Sequential(

    [

        tf.keras.layers.RandomRotation(
            0.08
        ),

        tf.keras.layers.RandomZoom(
            0.10
        ),

        tf.keras.layers.RandomTranslation(
            0.05,
            0.05
        ),

        tf.keras.layers.RandomContrast(
            0.10
        )

    ],

    name="tuning_augmentation"
)

print()
print("Data augmentation created.")

# CLASS WEIGHTS

print()
print("Calculating class weights...")

y_train = []

for images, labels in train_ds:

    labels_array = labels.numpy()

    label_indices = np.argmax(
        labels_array,
        axis=1
    )

    y_train.extend(
        label_indices.tolist()
    )

y_train = np.array(
    y_train
)

class_weights_array = (
    compute_class_weight(

        class_weight="balanced",

        classes=np.arange(
            len(CLASS_NAMES)
        ),

        y=y_train
    )
)

CLASS_WEIGHTS = {

    i: float(weight)

    for i, weight
    in enumerate(
        class_weights_array
    )
}

print()
print("Class weights:")

for i, class_name in enumerate(
    CLASS_NAMES
):

    print(
        f"{class_name}: "
        f"{CLASS_WEIGHTS[i]:.4f}"
    )

# CNN MODEL FUNCTION

def build_cnn(
    dropout_rate=0.40,
    learning_rate=0.0003
):

    inputs = tf.keras.Input(

        shape=(

            IMG_SIZE[0],

            IMG_SIZE[1],

            3

        )
    )

    # Augmentation
    x = augmentation(
        inputs
    )

    # Normalize
    x = tf.keras.layers.Rescaling(
        1.0 / 255.0
    )(x)

    # --------------------------------------------------------
    # Block 1
    # --------------------------------------------------------

    x = tf.keras.layers.Conv2D(
        32,
        (3, 3),
        padding="same"
    )(x)

    x = tf.keras.layers.BatchNormalization()(x)

    x = tf.keras.layers.Activation(
        "relu"
    )(x)

    x = tf.keras.layers.MaxPooling2D()(x)

    x = tf.keras.layers.SpatialDropout2D(
        0.10
    )(x)

    # --------------------------------------------------------
    # Block 2
    # --------------------------------------------------------

    x = tf.keras.layers.Conv2D(
        64,
        (3, 3),
        padding="same"
    )(x)

    x = tf.keras.layers.BatchNormalization()(x)

    x = tf.keras.layers.Activation(
        "relu"
    )(x)

    x = tf.keras.layers.MaxPooling2D()(x)

    x = tf.keras.layers.SpatialDropout2D(
        0.10
    )(x)

    # --------------------------------------------------------
    # Block 3
    # --------------------------------------------------------

    x = tf.keras.layers.Conv2D(
        128,
        (3, 3),
        padding="same"
    )(x)

    x = tf.keras.layers.BatchNormalization()(x)

    x = tf.keras.layers.Activation(
        "relu"
    )(x)

    x = tf.keras.layers.MaxPooling2D()(x)

    x = tf.keras.layers.SpatialDropout2D(
        0.15
    )(x)

    # --------------------------------------------------------
    # Block 4
    # --------------------------------------------------------

    x = tf.keras.layers.Conv2D(
        256,
        (3, 3),
        padding="same"
    )(x)

    x = tf.keras.layers.BatchNormalization()(x)

    x = tf.keras.layers.Activation(
        "relu"
    )(x)

    x = tf.keras.layers.MaxPooling2D()(x)

    x = tf.keras.layers.SpatialDropout2D(
        0.20
    )(x)

    # --------------------------------------------------------
    # Block 5
    # --------------------------------------------------------

    x = tf.keras.layers.Conv2D(
        512,
        (3, 3),
        padding="same"
    )(x)

    x = tf.keras.layers.BatchNormalization()(x)

    x = tf.keras.layers.Activation(
        "relu"
    )(x)

    # --------------------------------------------------------
    # Global Average Pooling
    # --------------------------------------------------------

    x = tf.keras.layers.GlobalAveragePooling2D()(x)

    # --------------------------------------------------------
    # Dense Layer
    # --------------------------------------------------------

    x = tf.keras.layers.Dense(
        256
    )(x)

    x = tf.keras.layers.BatchNormalization()(x)

    x = tf.keras.layers.Activation(
        "relu"
    )(x)

    x = tf.keras.layers.Dropout(
        dropout_rate
    )(x)

    # --------------------------------------------------------
    # Output
    # --------------------------------------------------------

    outputs = tf.keras.layers.Dense(

        len(CLASS_NAMES),

        activation="softmax"

    )(x)

    model = tf.keras.Model(

        inputs,

        outputs,

        name="tuned_cnn"
    )

    # --------------------------------------------------------
    # Optimizer
    # --------------------------------------------------------

    optimizer = (
        tf.keras.optimizers.AdamW(

            learning_rate=learning_rate,

            weight_decay=1e-4
        )
    )

    # --------------------------------------------------------
    # Loss
    # --------------------------------------------------------

    loss = (
        tf.keras.losses
        .CategoricalCrossentropy(

            label_smoothing=0.05
        )
    )

    # --------------------------------------------------------
    # Compile
    # --------------------------------------------------------

    model.compile(

        optimizer=optimizer,

        loss=loss,

        metrics=[
            "accuracy"
        ]
    )

    return model

#TEST MODEL CREATION

print()
print("Testing CNN model creation...")

test_model = build_cnn(

    dropout_rate=0.40,

    learning_rate=0.0003
)

print(
    "CNN model created successfully."
)

print()

test_model.summary()

del test_model

tf.keras.backend.clear_session()

#  DEFINE 3 HYPERPARAMETER EXPERIMENTS

experiments = [

    {
        "experiment": 1,
        "learning_rate": 0.001,
        "dropout": 0.30
    },

    {
        "experiment": 2,
        "learning_rate": 0.0003,
        "dropout": 0.40
    },

    {
        "experiment": 3,
        "learning_rate": 0.0001,
        "dropout": 0.50
    }

]

print()
print("=" * 70)
print("3 HYPERPARAMETER EXPERIMENTS")
print("=" * 70)

for experiment in experiments:

    print(

        f"Experiment "
        f"{experiment['experiment']}: "

        f"Learning Rate = "
        f"{experiment['learning_rate']}, "

        f"Dropout = "
        f"{experiment['dropout']}"

    )


# RUN 3 EXPERIMENTS


tuning_results = []


for experiment_info in experiments:

    experiment_number = (
        experiment_info[
            "experiment"
        ]
    )

    lr = (
        experiment_info[
            "learning_rate"
        ]
    )

    dropout = (
        experiment_info[
            "dropout"
        ]
    )

    print()
    print()
    print("=" * 70)

    print(
        f"EXPERIMENT "
        f"{experiment_number}/3"
    )

    print(
        f"Learning Rate : {lr}"
    )

    print(
        f"Dropout Rate  : {dropout}"
    )

    print("=" * 70)

    # Clear previous model
    tf.keras.backend.clear_session()

    # Reproducibility
    tf.keras.utils.set_random_seed(
        SEED + experiment_number
    )

    # Build model
    model = build_cnn(

        dropout_rate=dropout,

        learning_rate=lr
    )

    # Callbacks
    callbacks = [

        tf.keras.callbacks.EarlyStopping(

            monitor="val_accuracy",

            patience=PATIENCE,

            mode="max",

            restore_best_weights=True,

            verbose=1
        ),

        tf.keras.callbacks.ReduceLROnPlateau(

            monitor="val_loss",

            factor=0.5,

            patience=2,

            min_lr=1e-7,

            verbose=1
        )

    ]

    # Train
    history = model.fit(

        train_ds,

        validation_data=val_ds,

        epochs=EPOCHS,

        class_weight=CLASS_WEIGHTS,

        callbacks=callbacks,

        verbose=1
    )

    # Best validation accuracy
    best_val_accuracy = max(

        history.history[
            "val_accuracy"
        ]
    )

    # Best validation loss
    best_val_loss = min(

        history.history[
            "val_loss"
        ]
    )

    # Best epoch
    best_epoch = (

        np.argmax(

            history.history[
                "val_accuracy"
            ]

        ) + 1
    )

    # Store results
    tuning_results.append({

        "experiment":
            experiment_number,

        "learning_rate":
            lr,

        "dropout":
            dropout,

        "best_val_accuracy":
            best_val_accuracy,

        "best_val_loss":
            best_val_loss,

        "best_epoch":
            best_epoch

    })

    print()
    print(
        f"Experiment "
        f"{experiment_number} completed."
    )

    print(
        f"Best Validation Accuracy: "
        f"{best_val_accuracy:.4f}"
    )

    # Delete model before next experiment
    del model

    tf.keras.backend.clear_session()

# CREATE TUNING RESULTS DATAFRAME

tuning_df = pd.DataFrame(
    tuning_results
)

tuning_df = tuning_df.sort_values(

    by="best_val_accuracy",

    ascending=False

).reset_index(
    drop=True
)

print()
print()
print("=" * 70)
print("HYPERPARAMETER TUNING RESULTS")
print("=" * 70)

display(
    tuning_df
)

# SAVE TUNING RESULTS

tuning_csv_path = (

    RESULTS_DIR /

    "hyperparameter_tuning_results.csv"

)

tuning_df.to_csv(

    tuning_csv_path,

    index=False
)

print()
print(
    "Tuning results saved to:"
)

print(
    tuning_csv_path
)

# SELECT BEST HYPERPARAMETERS

best_row = (
    tuning_df.iloc[0]
)

BEST_LR = float(

    best_row[
        "learning_rate"
    ]
)

BEST_DROPOUT = float(

    best_row[
        "dropout"
    ]
)

BEST_VAL_ACCURACY = float(

    best_row[
        "best_val_accuracy"
    ]
)

print()
print()
print("=" * 70)
print("BEST HYPERPARAMETERS")
print("=" * 70)

print(
    f"Learning Rate      : "
    f"{BEST_LR}"
)

print(
    f"Dropout Rate       : "
    f"{BEST_DROPOUT}"
)

print(
    f"Validation Accuracy: "
    f"{BEST_VAL_ACCURACY:.4f}"
)

print()
print(
    f"Previous CNN Baseline: "
    f"{BASELINE_ACCURACY:.4f}"
)

# PLOT HYPERPARAMETER RESULTS


plt.figure(
    figsize=(10, 6)
)

labels = [

    f"Experiment {int(row.experiment)}"

    for _, row
    in tuning_df.iterrows()

]

values = (

    tuning_df[
        "best_val_accuracy"
    ].values

)

plt.bar(

    labels,

    values

)

plt.axhline(

    BASELINE_ACCURACY,

    linestyle="--",

    label="CNN Baseline"

)

plt.xlabel(
    "Experiment"
)

plt.ylabel(
    "Best Validation Accuracy"
)

plt.title(
    "Hyperparameter Tuning Results"
)

plt.legend()

plt.tight_layout()

tuning_plot_path = (

    FIGURES_DIR /

    "hyperparameter_tuning.png"

)

plt.savefig(

    tuning_plot_path,

    dpi=300,

    bbox_inches="tight"

)

plt.show()

print()
print(
    "Saved:"
)

print(
    tuning_plot_path
)

# BUILD FINAL TUNED CNN


print()
print(
    "Building final tuned CNN..."
)

tf.keras.backend.clear_session()

final_tuned_model = build_cnn(

    dropout_rate=BEST_DROPOUT,

    learning_rate=BEST_LR
)

print(
    "Final tuned CNN created."
)


# 22. FINAL MODEL PATH

TUNED_MODEL_PATH = (

    CNN_MODELS_DIR /

    "best_cnn_tuned.keras"

)

print()
print(
    "Final model will be saved to:"
)

print(
    TUNED_MODEL_PATH
)

# FINAL TRAINING CALLBACKS


final_callbacks = [

    tf.keras.callbacks.ModelCheckpoint(

        filepath=TUNED_MODEL_PATH,

        monitor="val_accuracy",

        mode="max",

        save_best_only=True,

        verbose=1

    ),

    tf.keras.callbacks.EarlyStopping(

        monitor="val_accuracy",

        patience=FINAL_PATIENCE,

        mode="max",

        restore_best_weights=True,

        verbose=1

    ),

    tf.keras.callbacks.ReduceLROnPlateau(

        monitor="val_loss",

        factor=0.3,

        patience=2,

        min_lr=1e-7,

        verbose=1

    )

]

# FINAL TRAINING

print()
print()
print("=" * 70)
print("FINAL TUNED CNN TRAINING")
print("=" * 70)

final_history = (

    final_tuned_model.fit(

        train_ds,

        validation_data=val_ds,

        epochs=FINAL_EPOCHS,

        class_weight=CLASS_WEIGHTS,

        callbacks=final_callbacks,

        verbose=1

    )

)

print()
print(
    "Final tuned CNN training completed."
)

print()
print(
    "Best model saved to:"
)

print(
    TUNED_MODEL_PATH
)

# LOAD BEST MODEL


print()
print(
    "Loading best tuned model..."
)

best_tuned_model = (

    tf.keras.models.load_model(

        TUNED_MODEL_PATH

    )

)

print(
    "Best tuned CNN loaded successfully."
)

# TEST EVALUATION


print()
print()
print("=" * 70)
print("TEST DATA EVALUATION")
print("=" * 70)

test_loss, test_accuracy = (

    best_tuned_model.evaluate(

        test_ds,

        verbose=1

    )

)

print()

print(
    f"Test Loss     : "
    f"{test_loss:.4f}"
)

print(
    f"Test Accuracy : "
    f"{test_accuracy:.4f}"
)

#  GENERATE PREDICTIONS


print()
print(
    "Generating predictions..."
)

y_true = []
y_pred = []

for images, labels in test_ds:

    predictions = (

        best_tuned_model.predict(

            images,

            verbose=0

        )

    )

    y_true.extend(

        np.argmax(

            labels.numpy(),

            axis=1

        )

    )

    y_pred.extend(

        np.argmax(

            predictions,

            axis=1

        )

    )

y_true = np.array(
    y_true
)

y_pred = np.array(
    y_pred
)

print(
    "Prediction completed."
)

print(
    "Number of test samples:",
    len(y_true)
)

#  CALCULATE METRICS


tuned_accuracy = (

    accuracy_score(

        y_true,

        y_pred

    )

)

tuned_precision = (

    precision_score(

        y_true,

        y_pred,

        average="weighted",

        zero_division=0

    )

)

tuned_recall = (

    recall_score(

        y_true,

        y_pred,

        average="weighted",

        zero_division=0

    )

)

tuned_f1 = (

    f1_score(

        y_true,

        y_pred,

        average="weighted",

        zero_division=0

    )

)

print()
print()
print("=" * 70)
print("TUNED CNN TEST METRICS")
print("=" * 70)

print(
    f"Accuracy : "
    f"{tuned_accuracy:.4f}"
)

print(
    f"Precision: "
    f"{tuned_precision:.4f}"
)

print(
    f"Recall   : "
    f"{tuned_recall:.4f}"
)

print(
    f"F1 Score : "
    f"{tuned_f1:.4f}"
)

# CLASSIFICATION REPORT


report = (

    classification_report(

        y_true,

        y_pred,

        target_names=CLASS_NAMES,

        digits=4,

        zero_division=0

    )

)

print()
print(
    "CLASSIFICATION REPORT"
)

print(
    report
)

report_path = (

    RESULTS_DIR /

    "tuned_cnn_classification_report.txt"

)

with open(

    report_path,

    "w",

    encoding="utf-8"

) as file:

    file.write(
        report
    )

print()
print(
    "Classification report saved:"
)

print(
    report_path
)

#  CONFUSION MATRIX


cm = confusion_matrix(

    y_true,

    y_pred

)

plt.figure(

    figsize=(8, 6)

)

sns.heatmap(

    cm,

    annot=True,

    fmt="d",

    xticklabels=CLASS_NAMES,

    yticklabels=CLASS_NAMES

)

plt.xlabel(
    "Predicted Label"
)

plt.ylabel(
    "True Label"
)

plt.title(
    "Tuned CNN Confusion Matrix"
)

plt.tight_layout()

cm_path = (

    FIGURES_DIR /

    "tuned_cnn_confusion_matrix.png"

)

plt.savefig(

    cm_path,

    dpi=300,

    bbox_inches="tight"

)

plt.show()

print()
print(
    "Confusion matrix saved:"
)

print(
    cm_path
)

#  BASELINE VS TUNED CNN

comparison_df = pd.DataFrame({

    "Model": [

        "CNN Baseline",

        "Tuned CNN"

    ],

    "Accuracy": [

        BASELINE_ACCURACY,

        tuned_accuracy

    ],

    "Precision": [

        BASELINE_PRECISION,

        tuned_precision

    ],

    "Recall": [

        BASELINE_RECALL,

        tuned_recall

    ],

    "F1 Score": [

        BASELINE_F1,

        tuned_f1

    ]

})

print()
print()
print("=" * 70)
print("BASELINE VS TUNED CNN")
print("=" * 70)

display(
    comparison_df
)

comparison_path = (

    RESULTS_DIR /

    "tuned_cnn_metrics.csv"

)

comparison_df.to_csv(

    comparison_path,

    index=False

)

print()
print(
    "Comparison saved:"
)

print(
    comparison_path
)

# ACCURACY IMPROVEMENT

accuracy_improvement = (

    tuned_accuracy -

    BASELINE_ACCURACY

)

accuracy_improvement_percentage = (

    accuracy_improvement * 100

)

print()
print(
    f"Baseline Accuracy : "
    f"{BASELINE_ACCURACY:.4f}"
)

print(
    f"Tuned Accuracy    : "
    f"{tuned_accuracy:.4f}"
)

print(
    f"Improvement       : "
    f"{accuracy_improvement_percentage:.2f} "
    f"percentage points"
)

#  TRAINING ACCURACY CURVE

plt.figure(

    figsize=(10, 6)

)

plt.plot(

    final_history.history[
        "accuracy"
    ],

    label="Training Accuracy"

)

plt.plot(

    final_history.history[
        "val_accuracy"
    ],

    label="Validation Accuracy"

)

plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "Accuracy"
)

plt.title(
    "Tuned CNN Training and Validation Accuracy"
)

plt.legend()

plt.grid(

    True,

    alpha=0.3

)

plt.tight_layout()

accuracy_curve_path = (

    FIGURES_DIR /

    "tuned_cnn_accuracy_curve.png"

)

plt.savefig(

    accuracy_curve_path,

    dpi=300,

    bbox_inches="tight"

)

plt.show()

print()
print(
    "Accuracy curve saved:"
)

print(
    accuracy_curve_path
)

#  TRAINING LOSS CURVE

plt.figure(

    figsize=(10, 6)

)

plt.plot(

    final_history.history[
        "loss"
    ],

    label="Training Loss"

)

plt.plot(

    final_history.history[
        "val_loss"
    ],

    label="Validation Loss"

)

plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "Loss"
)

plt.title(
    "Tuned CNN Training and Validation Loss"
)

plt.legend()

plt.grid(

    True,

    alpha=0.3

)

plt.tight_layout()

loss_curve_path = (

    FIGURES_DIR /

    "tuned_cnn_loss_curve.png"

)

plt.savefig(

    loss_curve_path,

    dpi=300,

    bbox_inches="tight"

)

plt.show()

print()
print(
    "Loss curve saved:"
)

print(
    loss_curve_path
)

# FINAL SUMMARY


print()
print()
print("=" * 70)
print("NOTEBOOK 06 COMPLETED SUCCESSFULLY")
print("=" * 70)

print()

print(
    "Best Learning Rate:"
)

print(
    BEST_LR
)

print()

print(
    "Best Dropout:"
)

print(
    BEST_DROPOUT
)

print()

print(
    "Best Validation Accuracy:"
)

print(
    f"{BEST_VAL_ACCURACY:.4f}"
)

print()

print(
    "Final Test Accuracy:"
)

print(
    f"{tuned_accuracy:.4f}"
)

print()

print(
    "Final Precision:"
)

print(
    f"{tuned_precision:.4f}"
)

print()

print(
    "Final Recall:"
)

print(
    f"{tuned_recall:.4f}"
)

print()

print(
    "Final F1 Score:"
)

print(
    f"{tuned_f1:.4f}"
)

print()
print(
    "Final Model:"
)

print(
    TUNED_MODEL_PATH
)

print()
print(
    "Tuning Results:"
)

print(
    tuning_csv_path
)

print()
print(
    "Classification Report:"
)

print(
    report_path
)

print()
print(
    "Confusion Matrix:"
)

print(
    cm_path
)

print()
print(
    "Accuracy Curve:"
)

print(
    accuracy_curve_path
)

print()
print(
    "Loss Curve:"
)

print(
    loss_curve_path
)

print()
print("=" * 70)
print("READY FOR NOTEBOOK 07 - GRAD-CAM")
print("=" * 70)
```
